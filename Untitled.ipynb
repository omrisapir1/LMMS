```python
import pandas as pd

# logs = [
# '2026-03-08T10:07:11 | update=1 | episodes=64 | tokens=2548 | reward_mean=0.1245 | exact=0.0469 | answer_rate=0.9531 | avg_len=39.81 | entropy=2.9932 | clipfrac=0.0035 | policy_loss=-0.6069 | value_loss=0.0651 | explained_var=0.0000 | rollouts=./runs/ppo/rollouts/rollouts_rank0_step0001.jsonl | t_sync=2.843s | t_rollout=4.764s | t_reward=0.003s | t_backprop=5.991s | t_update=13.607s',
# '2026-03-08T10:07:24 | update=2 | episodes=64 | tokens=2342 | reward_mean=0.2170 | exact=0.1250 | answer_rate=0.9531 | avg_len=36.59 | entropy=2.4863 | clipfrac=0.0050 | policy_loss=-0.8184 | value_loss=0.1449 | explained_var=-0.0033 | rollouts=./runs/ppo/rollouts/rollouts_rank0_step0002.jsonl | t_sync=2.149s | t_rollout=4.397s | t_reward=0.003s | t_backprop=6.387s | t_update=12.944s'
# ]

rows = []
for line in logs[1:-10]:
    parts = line.split(" | ")
    row = {"timestamp": parts[0]}
    for p in parts[1:]:
        k, v = p.split("=", 1)
        row[k] = v.rstrip("s")  # remove 's' from timing fields
    rows.append(row)

df = pd.DataFrame(rows)

# convert numeric columns when possible
df = df.apply(pd.to_numeric, errors="ignore")

print(df)
```
