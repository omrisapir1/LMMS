```python
from datasets import load_dataset

ds = load_dataset("omrisap/SFT_1024_40K")
```

```python
from datasets import load_dataset

ds_3 = load_dataset("omrisap/phase_3")
```

```python
df = ds['train'].to_pandas()
```

```python
ev_problems = set(ds_3['eval']['problem'])
```

```python
ev_problems
```

```python
remove_q = set(df[df['question'].isin(ev_problems)]['question'])
```

```python
df_eval = ds_3['eval'].to_pandas()
```

```python
df_eval = df_eval[~df_eval['problem'].isin(remove_q)]
```

```python

from datasets import Dataset
from huggingface_hub import login

# ====== CONFIG ======
HF_TOKEN = ""
REPO_ID = "omrisap/SFT_eval"  # e.g. "omrisap/my_dataset"
PRIVATE = False  # change to True if needed
# ====================

# 1) Login
login(token=HF_TOKEN)

# 2) Convert pandas -> HF Dataset
dataset = Dataset.from_pandas(df_eval, preserve_index=False,split='eval')

# 3) Push to Hub
dataset.push_to_hub(REPO_ID, private=PRIVATE)

print(f"✅ Dataset pushed to https://huggingface.co/datasets/{REPO_ID}")
```
